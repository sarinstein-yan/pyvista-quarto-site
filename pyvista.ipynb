## `PyVista` visualization of non-Hermiticity induced knotted graph topology transitions

When non-Hermitian perturbation is small, the original knot topology is preserved (@fig-th0.1). 

```python
#| label: fig-th0.1
#| fig-cap: Non-Hermiticity = 0.1. The original knot topology is preserved.
from IPython.display import HTML

def display_html(nonHermticity):
    ES_file = f"assets/ES_nH={nonHermticity}.html"
    SG_file = f"assets/SG_nH={nonHermticity}.html"
    # Display the saved HTML files side by side.
    return HTML(f"""
    <div style="display: flex; justify-content: space-between;">
        <iframe src="{ES_file}" width="800" height="500" style="border: none;"></iframe>
        <iframe src="{SG_file}" width="800" height="500" style="border: none;"></iframe>
    </div>
    """)

display_html(0.1)
```

When non-Hermiticity is prevalent enough, the exceptional surface starts to touch itself, leading to a topology transition --- the skeleton (i.e., medial axis) of the exceptional surface becomes a knotted graph (a.k.a. spatial multigraph).

As the non-Hermiticity evolves, the knotted graph topology evolves accordingly, leading to a plethora of exotic 3D spatial geometries (@fig-th0.2, @fig-th0.5).

```python
#| label: fig-th0.2
#| fig-cap: "Non-Hermiticity = 0.2. The exceptional surface starts to touch itself, and the skeleton of the exceptional surface becomes a knotted graph."

display_html(0.2)
```

```python
#| label: fig-th0.5
#| fig-cap: "Non-Hermiticity = 0.5. As non-Hermiticity evolves, the knotted graph topology evolves accordingly, leading to a plethora of exotic 3D spatial geometries."

display_html(0.5)
```
